```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
# ------------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------------
RAW_DATA_PATH = "used_cars.csv"
TARGET_COL = "price"
OUTPUT_TRAIN_PATH = "train.csv"
OUTPUT_TEST_PATH = "test.csv"
RANDOM_STATE = 42
TEST_SIZE = 0.2
HIGH_CARD_THRESHOLD = 50
```

```python
# ------------------------------------------------------------------
# STEP 1: Load
# ------------------------------------------------------------------
print("=" * 60)
print("STEP 1: Loading data")
print("=" * 60)

df = pd.read_csv(RAW_DATA_PATH)
print(f"Shape: {df.shape}")
print(df.dtypes)
```

```python
df.nunique()
```

```python
pd.set_option('display.max_columns', None)
df.head(15)
```

```python
# ------------------------------------------------------------------
# STEP 1b: Fix multi-word brand names that got mis-split at scrape time
# e.g. "Land" / "Rover Range Rover Sport..." -> "Land Rover" / "Range Rover Sport..."
# ------------------------------------------------------------------
MULTI_WORD_BRAND_FIXES = {
    "Land": ("Land Rover", "Rover"),
    "Aston": ("Aston Martin", "Martin"),
    "Alfa": ("Alfa Romeo", "Romeo"),
    "Rolls": ("Rolls-Royce", "Royce"),
}
if "brand" in df.columns and "model" in df.columns:
    for wrong_brand, (correct_brand, model_prefix) in MULTI_WORD_BRAND_FIXES.items():
        mask = df["brand"] == wrong_brand
        n = mask.sum()
        if n > 0:
            df.loc[mask, "model"] = (
                df.loc[mask, "model"].astype(str).str.replace(rf"^{model_prefix}\s*", "", regex=True)
            )
            df.loc[mask, "brand"] = correct_brand
            print(f"Fixed {n} rows: '{wrong_brand}' -> '{correct_brand}'")
```

```python
df.head(15)
```

```python
# ------------------------------------------------------------------
# STEP 2: Clean placeholder/encoding-artifact strings -> real NaN
# ------------------------------------------------------------------

# Tokens that represent "missing" but aren't real NaN (encoding artifacts, etc.)
PLACEHOLDER_TOKENS = {"â€“", "–", "-", "n/a", "na", "none", "unknown", ""}

def clean_placeholders(series):
    return series.apply(
        lambda x: np.nan if isinstance(x, str) and x.strip().lower() in PLACEHOLDER_TOKENS else x
    )

text_cols_to_clean = ["brand", "model", "fuel_type", "engine", "transmission", "ext_col", "int_col",
                       "accident", "clean_title"]
for col in text_cols_to_clean:
    if col in df.columns:
        df[col] = clean_placeholders(df[col])

print("\nNulls after placeholder cleanup:")
print(df[text_cols_to_clean].isnull().sum())
```

```python
# ------------------------------------------------------------------
# STEP 3: Parse text-encoded numeric columns
# ------------------------------------------------------------------
print("\n" + "=" * 60)
print("STEP 3: Parsing numeric-in-text columns")
print("=" * 60)

# price: "$10,300 " -> 10300.0  (TARGET)
df["price"] = (
    df["price"].astype(str).str.replace(r"[\$,]", "", regex=True).str.strip()
)
df["price"] = pd.to_numeric(df["price"], errors="coerce")

# milage: "51,000 mi." -> 51000.0
df["milage"] = (
    df["milage"].astype(str).str.replace(",", "", regex=False).str.extract(r"([\d.]+)")[0]
)
df["milage"] = pd.to_numeric(df["milage"], errors="coerce")

# --- Infer fuel_type = "Electric" when blank but engine text says so ---
df.loc[
    df["fuel_type"].isna() & df["engine"].str.contains("Electric", case=False, na=False),
    "fuel_type",
] = "Electric"

# --- Extract engine_hp, engine_liters, engine_cylinders from free text ---
def extract_hp(text):
    if pd.isna(text):
        return np.nan
    m = re.search(r"([\d.]+)\s*HP", str(text), flags=re.IGNORECASE)
    return float(m.group(1)) if m else np.nan

def extract_liters(text):
    if pd.isna(text):
        return np.nan
    # Format 1: "3.8L", "5.2L"
    m = re.search(r"([\d.]+)\s*L\b", str(text))
    if m:
        return float(m.group(1))
    # Format 2: "3.5 Liter", "2.0 Liter Supercharged"
    m = re.search(r"([\d.]+)\s*Liter\b", str(text), flags=re.IGNORECASE)
    if m:
        return float(m.group(1))
    return np.nan

def extract_cylinders(text):
    if pd.isna(text):
        return np.nan
    # Format 1: "6 Cylinder", "V6 Cylinder Engine", "Straight 6 Cylinder"
    m = re.search(r"(\d+)\s*Cylinder", str(text), flags=re.IGNORECASE)
    if m:
        return float(m.group(1))
    # Format 2: cylinder-config codes like "V6", "V-10", "I4", "H6", "W16"
    m = re.search(r"\b(?:V|I|H|W)-?(\d+)\b", str(text), flags=re.IGNORECASE)
    if m:
        return float(m.group(1))
    return np.nan

df["engine_hp"] = df["engine"].apply(extract_hp)
df["engine_liters"] = df["engine"].apply(extract_liters)
df["engine_cylinders"] = df["engine"].apply(extract_cylinders)

# Missing-indicator flags: missingness here is STRUCTURAL (EVs, bare "V6", "120 AH"
# battery specs, etc.), not random -> keep it visible to the model instead of letting
# imputation silently erase the distinction.
df["engine_hp_missing"] = df["engine_hp"].isna().astype(int)
df["engine_liters_missing"] = df["engine_liters"].isna().astype(int)
df["engine_cylinders_missing"] = df["engine_cylinders"].isna().astype(int)

print(f"engine_hp missing: {df['engine_hp_missing'].sum()} / {len(df)}")
print(f"engine_liters missing: {df['engine_liters_missing'].sum()} / {len(df)}")
print(f"engine_cylinders missing: {df['engine_cylinders_missing'].sum()} / {len(df)}")

df = df.drop(columns=["engine"])

# --- accident -> binary flag ---
# NOTE: blank/NaN is treated as "no accident reported" (same bucket as "Unknown").
# This is an assumption -- if you'd rather keep "unknown" separate from "confirmed
# none", change this to a 3-way category instead of collapsing to 0/1.
df["accident_reported"] = df["accident"].apply(
    lambda x: 1 if isinstance(x, str) and ("accident" in x.lower() or "damage" in x.lower()) else 0
)
df = df.drop(columns=["accident"])

# --- clean_title -> binary flag ("Yes" vs blank/anything else) ---
df["clean_title_flag"] = (df["clean_title"].astype(str).str.strip().str.lower() == "yes").astype(int)
df = df.drop(columns=["clean_title"])

# --- transmission: bucket into a small set of TYPES + extract SPEED COUNT ---
# Raw values (~61 unique) are mostly noise around a few real categories:
# "6-Speed A/T", "8-Speed Automatic", "A/T", "9-Speed Automatic with Auto-Shift"
# are all just "Automatic" with different wording/speed counts. One-hotting all
# 61 raw strings fragments one real signal across dozens of near-duplicate
# sparse columns. Splitting into type (low-cardinality categorical) + speed
# count (numeric) keeps the signal without the fragmentation.
def categorize_transmission(text):
    if pd.isna(text):
        return np.nan
    t = str(text).lower()
    if "cvt" in t or "variable" in t:
        return "CVT"
    if "dual shift" in t or "auto-shift" in t or "dct" in t:
        return "Automated Manual/DCT"
    if "m/t" in t or "manual" in t:
        return "Manual"
    if "a/t" in t or "automatic" in t or re.search(r"\bat\b", t):
        return "Automatic"
    return np.nan  # unrecognized (e.g. stray "F", "2") -> treated as missing

def extract_speeds(text):
    if pd.isna(text):
        return np.nan
    m = re.search(r"(\d+)\s*-?\s*Speed", str(text), flags=re.IGNORECASE)
    return float(m.group(1)) if m else np.nan

df["transmission_type"] = df["transmission"].apply(categorize_transmission)
df["transmission_speeds"] = df["transmission"].apply(extract_speeds)
df["transmission_speeds_missing"] = df["transmission_speeds"].isna().astype(int)

print(f"transmission_type nulls: {df['transmission_type'].isnull().sum()} / {len(df)}")
print(f"transmission_speeds nulls: {df['transmission_speeds'].isnull().sum()} / {len(df)}")

df = df.drop(columns=["transmission"])
```

```python
# ------------------------------------------------------------------
# STEP 4: Quick raw data-quality check (informs cleaning decisions below)
# No plots here -- duplicates/outliers haven't been removed yet, so a
# correlation heatmap or distribution plot at this point would be distorted.
# ------------------------------------------------------------------
print("\n" + "=" * 60)
print("STEP 4: Raw data-quality check (pre-cleaning)")
print("=" * 60)
print(df.describe())
print("\nMissing values per column:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df[TARGET_COL].dropna(), kde=True)
plt.title(f"Distribution of {TARGET_COL} (pre-cleaning)")
plt.tight_layout()
plt.savefig("eda_target_distribution_pre.png")
plt.show()
```

```python
# ------------------------------------------------------------------
# STEP 5: Clean data (dedupe, target validity, outliers)
# ------------------------------------------------------------------
print("\n" + "=" * 60)
print("STEP 5: Cleaning data")
print("=" * 60)
 
before = len(df)
df = df.drop_duplicates()
print(f"Dropped {before - len(df)} duplicate rows")
 
before = len(df)
df = df.dropna(subset=[TARGET_COL])
df = df[df[TARGET_COL] > 0]
print(f"Dropped {before - len(df)} rows with missing/invalid target")
 
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if TARGET_COL in numeric_cols:
    numeric_cols.remove(TARGET_COL)
categorical_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()
 
# 'model' is always dropped: hundreds of unique values, too granular to one-hot
# encode usefully. Its useful signal (luxury tier, performance trim, etc.) is
# already captured via 'brand' + the engine_hp/liters/cylinders features.
if "model" in categorical_cols:
    print(f"Dropping 'model' column ({df['model'].nunique()} unique values - too granular)")
    df = df.drop(columns=["model"])
    categorical_cols.remove("model")
 
# 'brand' is always KEPT regardless of cardinality -- it's a strong, meaningful
# predictor of price (a Bentley vs a Ford at the same mileage/year sells very
# differently), unlike 'model' which is closer to a near-unique identifier.
for col in list(categorical_cols):
    if col == "brand":
        continue
    n_unique = df[col].nunique()
    if n_unique > HIGH_CARD_THRESHOLD:
        print(f"Dropping high-cardinality column '{col}' ({n_unique} unique values)")
        df = df.drop(columns=[col])
        categorical_cols.remove(col)
 
print(f"\nNumeric columns: {numeric_cols}")
print(f"Categorical columns: {categorical_cols}")
 
def remove_outliers_iqr(data, columns, factor=1.5):
    data = data.copy()
    for col in columns:
        Q1, Q3 = data[col].quantile(0.25), data[col].quantile(0.75)
        IQR = Q3 - Q1
        lower, upper = Q1 - factor * IQR, Q3 + factor * IQR
        before_n = len(data)
        data = data[(data[col] >= lower) & (data[col] <= upper)]
        if before_n - len(data) > 0:
            print(f"Removed {before_n - len(data)} outliers on '{col}' (bounds: [{lower:.2f}, {upper:.2f}])")
    return data
 
# Only prune outliers on core, mostly-complete numeric columns.
# (Deliberately NOT pruning on engine_hp/liters/cylinders -- too many
# structurally-missing values there would make IQR bounds unreliable.)
outlier_check_cols = [c for c in [TARGET_COL, "milage", "model_year"] if c in df.columns]
df = remove_outliers_iqr(df, outlier_check_cols)
df = df[(df["model_year"] >= 1980) & (df["model_year"] <= 2026)]
 
print(f"\nShape after cleaning: {df.shape}")
```

```python
# ------------------------------------------------------------------
# STEP 5b: EDA plots on the CLEANED dataset
# Now that duplicates, invalid/missing targets, and outliers are gone, these
# plots reflect the actual distribution/correlations of the data that will
# feed the model -- not distorted by dupes or the $1.9M-outlier-type rows.
# ------------------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.histplot(df[TARGET_COL].dropna(), kde=True)
plt.title(f"Distribution of {TARGET_COL} (post-cleaning)")
plt.tight_layout()
plt.savefig("eda_target_distribution.png")
plt.show()
```

```python
# ------------------------------------------------------------------
# STEP 6: Preprocessing pipeline
# (All imputation happens HERE, fit on train only -> no leakage)
# ------------------------------------------------------------------
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_cols),
    ("cat", categorical_transformer, categorical_cols),
])
```

```python
# ------------------------------------------------------------------
# STEP 7: Split 80/20
# ------------------------------------------------------------------
X = df[numeric_cols + categorical_cols]
y = df[TARGET_COL]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE
)
print(f"Train shape: {X_train.shape} | Test shape: {X_test.shape}")

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

cat_feature_names = list(
    preprocessor.named_transformers_["cat"]["onehot"].get_feature_names_out(categorical_cols)
)
all_feature_names = numeric_cols + cat_feature_names

train_df = pd.DataFrame(X_train_processed, columns=all_feature_names)
test_df = pd.DataFrame(X_test_processed, columns=all_feature_names)
```

```python
# ------------------------------------------------------------------
# STEP 8: Save CSVs (SageMaker XGBoost format: target first, no header)
# ------------------------------------------------------------------
train_out = pd.concat([y_train.reset_index(drop=True), train_df.reset_index(drop=True)], axis=1)
test_out = pd.concat([y_test.reset_index(drop=True), test_df.reset_index(drop=True)], axis=1)

train_out.to_csv(OUTPUT_TRAIN_PATH, index=False, header=False)
test_out.to_csv(OUTPUT_TEST_PATH, index=False, header=False)
train_out.to_csv("train_with_headers.csv", index=False)
test_out.to_csv("test_with_headers.csv", index=False)

print(f"\nSaved: {OUTPUT_TRAIN_PATH}, {OUTPUT_TEST_PATH} (SageMaker XGBoost format)")
print("Saved: train_with_headers.csv / test_with_headers.csv (for review)")
print("\nDone! Upload train.csv and test.csv to your S3 bucket for Part B.")
```
